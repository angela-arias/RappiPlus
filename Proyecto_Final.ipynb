El objetivo de este proyecto es evaluar el desempeño del servicio **RappiPlus** para apoyar **decisiones de negocio basadas en datos**.

```python
# comparar floats con == suele dar falsos negativos por errores de redondeo (por ejemplo, 0.1 + 0.2 != 0.3).
# Si monto_total y monto_calculado tienen decimales, es más seguro usar una tolerancia, dataset orders:

orders["coinciden"] = np.isclose(
    orders["monto_calculado"], orders["monto_total"], atol=0.01
).astype(int)
```

```python
#Para revisar las filas que no coinciden y ver de cuánto es la diferencia, dataset orders:
orders["diferencia"] = orders["monto_calculado"] - orders["monto_total"]
print(orders[orders["coinciden"] == 0][["monto_calculado", "monto_total", "diferencia"]])
```

Las 50 filas que aparecen corresponden a las que anteriormente vimos que tenian cantidad, precio_unitario y monto_descuento en NaN.

```python
# Se  excluyen las filas anteriores, dataset orders:
validas = orders["monto_calculado"].notna()

orders["coinciden"] = np.nan
orders.loc[validas, "coinciden"] = np.isclose(
    orders.loc[validas, "monto_calculado"],
    orders.loc[validas, "monto_total"],
    atol=0.01
).astype(int)
```

```python
# Mostrar resultado, dataset orders:
print(orders["coinciden"].value_counts(dropna=False))
```

24946 filas coinciden (1.0). 50 no se pueden verificar (NaN): el bloque sin cantidad, precio ni descuento.

Se hizo la verificación de la consistencia de los montos. En 24946 filas, el monto calculado (cantidad × precio_unitario − monto_descuento) coincide con monto_total (tolerancia de 0.01). 50 filas no se pudieron verificar (NaN) porque no tienen cantidad, precio ni descuento; se conservan por tener monto_total válido.
```python
# Verificación de Outliers en Orders:
Q1_orders = orders["monto_total"].quantile(0.25)
Q3_orders = orders["monto_total"].quantile(0.75)
```

```python
# IQR dataset orders:
IQR_orders = Q3_orders - Q1_orders
```

```python
# Calcular los límites inferior y superior para detectar los outlier dataset orderss:
limite_inferior_orders = Q1_orders - 1.5 * IQR_orders
limite_superior_orders = Q3_orders + 1.5 * IQR_orders
```

```python
# Identificar cuántas filas de orders caen fuera de ese rango dataset orders:
outliers_orders = orders[(orders["monto_total"] < limite_inferior_orders)  | (orders["monto_total"] > limite_superior_orders)]
```

```python
# Verificación que monto_total sea consistente con esa cantidad dataset orders:
print(outliers_orders[["cantidad", "precio_unitario", "monto_descuento", "monto_total"]]
      .assign(calculado=lambda d: d["cantidad"] * d["precio_unitario"] - d["monto_descuento"]))
```

```python
# PARTE 1: Totales del funnel
# ======================
query_funnel = '''
WITH first_visit AS (
    SELECT DISTINCT id_usuario
    FROM events
    WHERE nombre_evento = 'first_visit'
),
select_item AS (
    SELECT DISTINCT e.id_usuario
    FROM events e
    JOIN first_visit f ON e.id_usuario = f.id_usuario
    WHERE e.nombre_evento = 'select_item'
),
add_to_cart AS (
    SELECT DISTINCT e.id_usuario
    FROM events e
    JOIN select_item s ON e.id_usuario = s.id_usuario
    WHERE e.nombre_evento = 'add_to_cart'
),
begin_checkout AS (
    SELECT DISTINCT e.id_usuario
    FROM events e
    JOIN add_to_cart a ON e.id_usuario = a.id_usuario
    WHERE e.nombre_evento = 'begin_checkout'
),
add_payment_info AS (
    SELECT DISTINCT e.id_usuario
    FROM events e
    JOIN begin_checkout b ON e.id_usuario = b.id_usuario
    WHERE e.nombre_evento = 'add_payment_info'
),
purchase AS (
    SELECT DISTINCT e.id_usuario
    FROM events e
    JOIN add_payment_info p ON e.id_usuario = p.id_usuario
    WHERE e.nombre_evento = 'purchase'
)
SELECT 1 AS orden, 'first_visit' AS nombre_evento, COUNT(*) AS total_usuarios FROM first_visit
UNION ALL SELECT 2, 'select_item', COUNT(*) FROM select_item
UNION ALL SELECT 3, 'add_to_cart', COUNT(*) FROM add_to_cart
UNION ALL SELECT 4, 'begin_checkout', COUNT(*) FROM begin_checkout
UNION ALL SELECT 5, 'add_payment_info', COUNT(*) FROM add_payment_info
UNION ALL SELECT 6, 'purchase', COUNT(*) FROM purchase
ORDER BY orden;
'''

funnel = pd.read_sql(query_funnel, con=engine)
funnel
```

El funnel se construyó de forma secuencial con CTEs: cada etapa incluye solo a los usuarios que también estuvieron en la etapa anterior. De 7.796 usuarios con first_visit, 7.393 llegan a select_item, 7.052 a add_to_cart, 6.364 a begin_checkout, 4.967 a add_payment_info y 3.857 a purchase. Con este criterio ninguna etapa supera a la anterior.

```python
# PARTE 2: Conversiones
funnel["usuarios_anterior"] = funnel["total_usuarios"].shift(1)
funnel["conversion_paso_pct"] = (funnel["total_usuarios"] / funnel["usuarios_anterior"] * 100).round(2)
funnel["usuarios_perdidos"] = funnel["usuarios_anterior"] - funnel["total_usuarios"]
funnel["perdida_pct"] = (100 - funnel["conversion_paso_pct"]).round(2)
print(funnel)

# Mayor pérdida (se excluye la primera fila, que no tiene etapa anterior)
peor = funnel.iloc[1:].loc[funnel.iloc[1:]["perdida_pct"].idxmax()]
print(f"Mayor pérdida: {peor['nombre_evento']} ({int(peor['usuarios_perdidos'])} usuarios, {peor['perdida_pct']}%)")

# Conversión final: compradores / visitantes
visitas = funnel.loc[funnel["nombre_evento"] == "first_visit", "total_usuarios"].iloc[0]
compras = funnel.loc[funnel["nombre_evento"] == "purchase", "total_usuarios"].iloc[0]
print(f"Conversión final (purchase / first_visit): {compras / visitas * 100:.2f}%")
```

Las conversiones entre pasos son 94,83%, 95,39%, 90,24%, 78,05% y 77,65%. Las mayores pérdidas se concentran en el tramo final: begin_checkout → add_payment_info pierde más usuarios en términos absolutos (1.397; 21,95%) y add_payment_info → purchase tiene la mayor pérdida porcentual (1.110 usuarios; 22,35%). Ambos pasos son los principales puntos de abandono.

```python
# Verificación del orden temporal: usuarios cuyo primer add_to_cart es anterior a su primer select_item.
# El funnel se construye por presencia secuencial de etapas; esta revisión muestra que las fechas
# de los eventos no siempre siguen el orden del recorrido.

query_orden = '''
SELECT COUNT(*) AS usuarios_fuera_de_orden
FROM (
    SELECT id_usuario,
           MIN(CASE WHEN nombre_evento = 'select_item' THEN timestamp_evento END) AS ts_select,
           MIN(CASE WHEN nombre_evento = 'add_to_cart' THEN timestamp_evento END) AS ts_cart
    FROM events
    GROUP BY id_usuario
) t
WHERE ts_cart < ts_select;
'''
pd.read_sql(query_orden, con=engine)
```

Se encontraron 3.610 usuarios cuyo primer add_to_cart tiene una fecha anterior a su primer select_item. Como timestamp_evento solo contiene fecha (sin hora), los eventos del mismo día no se consideran fuera de orden. Dado que una parte importante de los usuarios no respeta el orden cronológico esperado, el funnel se definió como secuencial por presencia de etapas (cada etapa parte de los usuarios de la etapa anterior) y no por orden de fechas. ## 🔹 Paso 4: Evaluar si los usuarios regresan (retención por cohortes)

Cada usuario tiene un registro en los días 7, 14, 21 y 28 (32000 filas, sin duplicados). activo solo toma los valores 0 y 1. Se calculan las semanas 1 a 3 según las instrucciones; el día 28 queda fuera del análisis.

```python
query_cohorte_usuario = '''
SELECT
    id_usuario,
    CAST(fecha_registro AS DATE) AS fecha_registro,
    TO_CHAR(DATE_TRUNC('month', CAST(fecha_registro AS DATE)), 'YYYY-MM') AS cohorte
FROM users
ORDER BY id_usuario;
'''

cohorte_usuario = pd.read_sql(query_cohorte_usuario, con=engine)
cohorte_usuario.head(10)
```

```python
# Resumen por cohorte:
print(cohorte_usuario["cohorte"].value_counts().sort_index())
```

La cohorte de cada usuario se identifica por el mes de fecha_registro (convertida a DATE con CAST). Hay cinco cohortes, de 2025-01 a 2025-05, que suman 8000 usuarios.

```python
# Retención acumulada por cohorte de registro (mes de fecha_registro).
# semana_num = dias_despues_registro / 7 (valores 1 a 4).
# retenido_wN: usuarios únicos con activo = 1 y semana_num >= N (actividad en la semana N o posterior).
# semana_N: retenido_wN dividido entre el tamaño de la cohorte (todos los usuarios registrados ese mes).

# Retención por cohortes
# ======================

query_cohorte_final = '''
WITH cohorte AS (
    SELECT
        id_usuario,
        TO_CHAR(DATE_TRUNC('month', CAST(fecha_registro AS DATE)), 'YYYY-MM') AS cohorte
    FROM users
),
tamano AS (
    SELECT cohorte, COUNT(*) AS usuarios_cohorte
    FROM cohorte
    GROUP BY cohorte
),
actividad AS (
    SELECT
        ua.id_usuario,
        c.cohorte,
        ua.dias_despues_registro / 7 AS semana_num,
        ua.activo
    FROM user_activity ua
    JOIN cohorte c ON ua.id_usuario = c.id_usuario
),
retenidos AS (
    SELECT
        cohorte,
        COUNT(DISTINCT CASE WHEN semana_num >= 1 AND activo = 1 THEN id_usuario END) AS retenido_w1,
        COUNT(DISTINCT CASE WHEN semana_num >= 2 AND activo = 1 THEN id_usuario END) AS retenido_w2,
        COUNT(DISTINCT CASE WHEN semana_num >= 3 AND activo = 1 THEN id_usuario END) AS retenido_w3
    FROM actividad
    GROUP BY cohorte
)
SELECT
    t.cohorte,
    t.usuarios_cohorte,
    r.retenido_w1,
    r.retenido_w2,
    r.retenido_w3,
    ROUND(100.0 * r.retenido_w1 / t.usuarios_cohorte, 2) AS semana_1,
    ROUND(100.0 * r.retenido_w2 / t.usuarios_cohorte, 2) AS semana_2,
    ROUND(100.0 * r.retenido_w3 / t.usuarios_cohorte, 2) AS semana_3
FROM tamano t
JOIN retenidos r ON t.cohorte = r.cohorte
ORDER BY t.cohorte;
'''

cohorte_final = pd.read_sql(query_cohorte_final, con=engine)
cohorte_final
```

```python
# Verificaciones: la suma de las cohortes debe ser 8000 (total de users) y,
# al ser retención acumulada, semana_1 >= semana_2 >= semana_3 en todas las cohortes.
print(cohorte_final["usuarios_cohorte"].sum())
print((cohorte_final["semana_1"] >= cohorte_final["semana_2"]).all(),
      (cohorte_final["semana_2"] >= cohorte_final["semana_3"]).all())
```

La retención se calculó de forma acumulada por cohorte de registro (mes de fecha_registro, convertida a DATE con CAST). La semana se obtiene como dias_despues_registro / 7, y un usuario cuenta como retenido en la semana N si estuvo activo (activo = 1) en la semana N o en una posterior (semana_num >= N). El denominador es el tamaño de la cohorte, es decir, todos los usuarios registrados ese mes.

La retención en la semana 1 es de ~85% a ~87% en todas las cohortes, baja a ~77% a ~80% en la semana 2 y a ~63% a ~65% en la semana 3. Las cinco cohortes (enero a mayo) suman 8.000 usuarios y su comportamiento es muy parecido, sin una tendencia clara entre meses.

```python
# Verificaciones:
print(experiment["variante"].value_counts())
print(experiment["convirtio"].value_counts(dropna=False))
print(experiment["id_usuario"].duplicated().sum())
```

```python
# Revisión de usuarios unicos:
print(experiment["id_usuario"].nunique())
print(len(experiment))
```

El experimento tiene 10000 usuarios únicos (uno por fila), sin duplicados, por lo que ningún usuario aparece en ambos grupos. Reparto: tratamiento 5035 y control 4965.

La columna de grupo es variante (control y tratamiento) y la métrica de conversión es convirtio (1 si el usuario completó la compra, 0 si no). Las instrucciones la llaman conversion; en el archivo el nombre es convirtio.

```python
# Tabla por grupo: usuarios, conversiones y tasa de conversión:
tabla_grupos = experiment.groupby("variante")["convirtio"].agg(
    usuarios="count",
    conversiones="sum",
    tasa="mean"
)
tabla_grupos["tasa_pct"] = (tabla_grupos["tasa"] * 100).round(2)
print(tabla_grupos)
```

```python
# Hipótesis y test:
from statsmodels.stats.proportion import proportions_ztest, proportion_confint

# Orden: control primero, tratamiento después
conversiones = [tabla_grupos.loc["control", "conversiones"],
                tabla_grupos.loc["tratamiento", "conversiones"]]
usuarios = [tabla_grupos.loc["control", "usuarios"],
            tabla_grupos.loc["tratamiento", "usuarios"]]

alpha = 0.05
z_stat, p_value = proportions_ztest(count=conversiones, nobs=usuarios, alternative="two-sided")

print(f"z = {z_stat:.4f}")
print(f"p-value = {p_value:.4f}")
print("Se rechaza H0" if p_value < alpha else "No se rechaza H0")
```

```python
# Intervalo de confianza:

p_c = tabla_grupos.loc["control", "tasa"]
p_t = tabla_grupos.loc["tratamiento", "tasa"]
n_c = tabla_grupos.loc["control", "usuarios"]
n_t = tabla_grupos.loc["tratamiento", "usuarios"]

dif = p_t - p_c
ee = np.sqrt(p_c * (1 - p_c) / n_c + p_t * (1 - p_t) / n_t)
ic_inf, ic_sup = dif - 1.96 * ee, dif + 1.96 * ee

print(f"Diferencia: {dif*100:.2f} pp")
print(f"IC 95%: [{ic_inf*100:.2f}, {ic_sup*100:.2f}] pp")
```

```python
# La tasa de conversión con intervalo de confianza:

grupos = ["control", "tratamiento"]
tasas = [tabla_grupos.loc[g, "tasa"] * 100 for g in grupos]
errores = [1.96 * np.sqrt(tabla_grupos.loc[g, "tasa"] * (1 - tabla_grupos.loc[g, "tasa"])
                          / tabla_grupos.loc[g, "usuarios"]) * 100 for g in grupos]

plt.figure(figsize=(7, 5))
barras = plt.bar(grupos, tasas, yerr=errores, capsize=8, color=["#95D5B2", "#F4A261"])

for barra, tasa in zip(barras, tasas):
    plt.text(barra.get_x() + barra.get_width() / 2, tasa / 2, f"{tasa:.2f}%",
             ha="center", va="center", fontsize=12, fontweight="bold")

plt.title("Tasa de conversión por variante (IC 95%)", fontsize=15)
plt.xlabel("Variante", fontsize=12)
plt.ylabel("Conversión (%)", fontsize=12)
plt.ylim(0, 20)
plt.show()
```

Las tasas de conversión de control (15.69%) y tratamiento (16.29%) son similares y sus intervalos de confianza al 95% se solapan. El test z (α = 0.05, p = 0.4161) y el intervalo de la diferencia [-0.84, 2.03] pp, que incluye el 0, indican que no hay evidencia estadística de que la nueva UI modifique la conversión.

No rechazar H₀ no demuestra que ambas UI sean iguales: significa que con estos datos no se detectó un efecto.

## Limitaciones
